```python

import numpy as np
arr = np.arange(20)
print("Каждый 3-й с индекса 1:", arr[1::3])
print("С 5-го по 15-й в обратном порядке:", arr[14:4:-1])
print("Последние 5 в обратном порядке:", arr[-1:-6:-1], "\n")

np.random.seed(42)
mat = np.random.randint(0, 10, size=(6, 6))
print("Матрица 6x6:\n", mat)
print("Подматрица строки 2-4, столбцы 1-3:\n", mat[2:5, 1:4])
print("Главная диагональ:", np.diagonal(mat))

idx_sum = np.arange(6)[:, None] + np.arange(6)
print("Шахматный узор (элементы):\n", mat[idx_sum % 2 == 0], "\n")
a = np.array([10, 20, 30, 40, 50, 60])
m = np.arange(12).reshape(3, 4)
print("Элементы a [0, 2, 4]:", a[[0, 2, 4]])
print("Попарно (0,1), (1,3), (2,0):", m[[0, 1, 2], [1, 3, 0]])
print("Блок 2x2 из строк [0, 2] и столбцов [1, 3]:\n", m[np.ix_([0, 2], [1, 3])], "\n")

x = np.random.randn(1000)
print("Количество элементов > 1:", np.sum(x > 1))

x_no_neg = x.copy()
x_no_neg[x_no_neg < 0] = 0

x_zero_cross = x.copy()
sign_change = (x[:-2] * x[2:]) < 0
x_zero_cross[1:-1][sign_change] = 0
print("Обнулено элементов при смене знака соседей:", np.sum(sign_change))


```
