```python
import os
import time
shape = (50000, 200)
fp = np.memmap('big.dat', dtype='float32', mode='w+', shape=shape)
fp[:] = np.random.rand(*shape).astype('float32')
fp.flush()
del fp

file_size = os.path.getsize('big.dat')
expected_size = 50000 * 200 * 4
print(f"7.1 Размер файла: {file_size} байт (Ожидалось: {expected_size})")

fp_read = np.memmap('big.dat', dtype='float32', mode='r', shape=shape)
batch_size = 5000
global_min = np.inf
global_max = -np.inf
total_sum = 0.0
total_count = 0

for i in range(0, shape[0], batch_size):
    batch = fp_read[i:i + batch_size]
    global_min = min(global_min, float(batch.min()))
    global_max = max(global_max, float(batch.max()))
    total_sum += float(batch.sum())
    total_count += batch.size

print(f"7.2 Min: {global_min:.4f}, Max: {global_max:.4f}, Mean: {total_sum / total_count:.4f}\n")
del fp_read

N = 3000
arr_c = np.ones((N, N), order='C')
arr_f = np.ones((N, N), order='F')

def sum_rows(a):
    s = 0.0
    for i in range(a.shape[0]):
        for j in range(a.shape[1]):
            s += a[i, j]
    return s

def sum_cols(a):
    s = 0.0
    for j in range(a.shape[1]):
        for i in range(a.shape[0]):
            s += a[i, j]
    return s

t0 = time.time(); sum_rows(arr_c); print(f"7.3 C-order построчно: {time.time() - t0:.3f} с")
t0 = time.time(); sum_cols(arr_c); print(f"7.3 C-order постолбцово: {time.time() - t0:.3f} с")
t0 = time.time(); sum_rows(arr_f); print(f"7.3 F-order построчно: {time.time() - t0:.3f} с")
t0 = time.time(); sum_cols(arr_f); print(f"7.3 F-order постолбцово: {time.time() - t0:.3f} с\n")

M = 10_000_000
a = np.random.rand(M)
b = np.random.rand(M)
a_list, b_list = a.tolist(), b.tolist()

t0 = time.time()
res_list = [(x + y) / (x - y + 1.0) for x, y in zip(a_list, b_list)]
print(f"7.4 Python list: {time.time() - t0:.3f} с")

t0 = time.time()
res_np = (a + b) / (a - b + 1.0)
print(f"7.4 NumPy 1-line: {time.time() - t0:.4f} с")

t0 = time.time()
num = np.empty_like(a)
den = np.empty_like(a)
np.add(a, b, out=num)
np.subtract(a, b, out=den)
np.add(den, 1.0, out=den)
np.divide(num, den, out=num)
print(f"7.4 NumPy с out=: {time.time() - t0:.4f} с")

```
